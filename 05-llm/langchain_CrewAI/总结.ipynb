```python
from crewai import Agent, Task, Crew, Process
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

llm = ChatOpenAI(
    model="GLM-5.3",
    temperature=0.9,
    api_key=os.getenv("GLM_API_KEY"),
    base_url=os.getenv("GLM_base_url")
)

# 定义agent角色
researcher = Agent(
    role="市场研究员",
    goal="收集并整理关于目标主题的全面、准确市场信息",
    backstory="你是一个经验丰富的市场分析师，擅长从海量信息中提炼关键",
    llm=llm
)
```

crewai的Agent(llm=)不兼容ChatOpenAI的llm

报错的原因很直接：CrewAI 的 Agent 不接受 langchain_openai.ChatOpenAI 对象作为 llm 参数。

Input should be a valid dictionary or instance of BaseLLM [type=model_type, input_value=ChatOpenAI(metadata={'lc_...eam_chunk_timeout=120.0), input_type=ChatOpenAI]
    For further information visit https://errors.pydantic.dev/2.12/v/model_type
